```python
# Import library
import librosa
import numpy as np
import soundfile as sf
from pydub import AudioSegment
```

```python
SAMPLE_WAV_FILE = "dataset/sample.wav"
```

```python
# pitch shift & white noise
audio, sample_rate = librosa.load(SAMPLE_WAV_FILE)
# repeat and fade
clip = AudioSegment.from_wav(SAMPLE_WAV_FILE)
```

```python
def get_filename():
    filename = input("New filename (.wav): ")
    
    while not filename.endswith(".wav"):
        filename = input("Filename must end with .wav. New filename (.wav): ")
        
    return filename
```

```python
def pitch_shift():
    while True:
        try:
            value = float(input("Enter pitch shift [-5 to 5]: "))
            if -5 <= value <= 5:
                break
        except ValueError:
            pass
        print("Invalid input. Please enter a number between -5 and 5.")
        
    filename = get_filename()
    result = librosa.effects.pitch_shift(audio, sample_rate, n_steps=value)
    sf.write(filename, result, sample_rate)
    print("Saved: ", filename)
```
